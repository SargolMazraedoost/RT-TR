# check duplication of two dataset of smiles

```python
import glob

import pandas as pd
from rdkit import Chem
```

```python
# Function to canonicalize a SMILES string
def canonicalize_smiles(smiles):
    try:
        mol = Chem.MolFromSmiles(smiles)
        if mol:
            return Chem.MolToSmiles(mol, canonical=True)
    except:
        return None  # Return None for invalid SMILES
    return None


# Load and canonicalize reference SMILES
smrt_data = pd.read_csv("../../data/SMRT_dataset.txt", sep="\t")
smrt_data["canonical_smiles"] = smrt_data["smiles"].apply(canonicalize_smiles)
ref_column = smrt_data["canonical_smiles"].dropna()

# Initialize a list to store file name and duplicate count
duplicate_summary = []

paths = glob.glob("../../data/dataset/10_subdataset_smiles/*.txt")

# Check if each data file has SMILES that are also in the reference SMILES
for path in paths:
    name = path.split("/")[-1]  # Extract the file name
    print("File name:", name)

    data = pd.read_csv(path, sep="\t")
    data = data.dropna()

    # Canonicalize SMILES in the current file
    data["canonical_smiles"] = data["smiles"].apply(canonicalize_smiles)

    # Count the matching SMILES in the reference
    matching_count = sum(data["canonical_smiles"].isin(ref_column))

    # Append the file name and duplicate count to the summary list
    duplicate_summary.append({"File Name": name, "Duplicate Count": matching_count})

    # Print matching SMILES if you want to see them for each file
    print("Number of SMILES in both datasets:", matching_count) if len(data) > 0 else None
    print("Matching SMILES:")
    print(data[data["canonical_smiles"].isin(ref_column)])
    print("=====================================")

# Convert the summary list to a DataFrame for easy viewing or saving
duplicate_summary_df = pd.DataFrame(duplicate_summary)

# Display the summary table
print(duplicate_summary_df)

# Optionally, save the summary to a file
duplicate_summary_df.to_csv("../../data/duplicate_summary.csv", index=False)
```
