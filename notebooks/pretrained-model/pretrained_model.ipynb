# Save the tokenizer and model to the specified directory

```python
from transformers import AutoConfig, AutoModelForSequenceClassification, AutoTokenizer


use_pretrained = True  # Change this to False for random initialization

# Define the checkpoint and derive a name for the model directory.
checkpoint = "DeepChem/ChemBERTa-10M-MTR"

name = checkpoint.split("/")[1]
save_model_path = f"../../models/{name}_{'pretrained' if use_pretrained else 'random'}"

# Load the tokenizer from the checkpoint.
tokenizer = AutoTokenizer.from_pretrained(checkpoint)

# Load or initialize the model based on the use_pretrained flag.
if use_pretrained:
    # Load a pretrained model (including its weights) from the checkpoint.
    model = AutoModelForSequenceClassification.from_pretrained(
        checkpoint,
        ignore_mismatched_sizes=True,
        num_labels=1,
    )
    print("Loaded pretrained model weights.")
else:
    config = AutoConfig.from_pretrained(checkpoint)
    config.num_labels = 1
    # Initialize the model from the configuration with random weights.
    model = AutoModelForSequenceClassification.from_config(config)
    print("Initialized model with random weights.")

# Optionally, set the pad token for both tokenizer and model if not already defined.
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
if model.config.pad_token_id is None:
    model.config.pad_token_id = tokenizer.pad_token_id

# Save the tokenizer and model to the specified directory.
tokenizer.save_pretrained(save_model_path)
model.save_pretrained(save_model_path)

print(f"Model and tokenizer saved to {save_model_path}")
```
